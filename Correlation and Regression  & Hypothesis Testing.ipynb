```python
import numpy as np
from scipy.stats import pearsonr
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

# --------------------------------------------------
# 1. Data
# --------------------------------------------------

# Independent variable: Hours Studied
X = np.array([2, 3, 4, 5, 6, 7, 8, 9, 10])

# Dependent variable: Exam Score
Y = np.array([50, 55, 65, 60, 70, 75, 78, 85, 90])

# --------------------------------------------------
# 2. Pearson Correlation
# --------------------------------------------------

correlation, p_value = pearsonr(X, Y)

print("----- Pearson Correlation -----")
print("Correlation coefficient (r):", round(correlation, 4))
print("p-value:", format(p_value, ".6f"))

# --------------------------------------------------
# 3. Hypothesis Testing
# --------------------------------------------------

alpha = 0.05

print("\n----- Hypothesis Testing -----")
print("H0: There is no significant linear correlation.")
print("H1: There is a significant linear correlation.")

if p_value < alpha:
    print("Decision: Reject H0")
    print("Conclusion: The correlation is statistically significant.")
else:
    print("Decision: Fail to reject H0")
    print("Conclusion: The correlation is not statistically significant.")

# --------------------------------------------------
# 4. Simple Linear Regression
# --------------------------------------------------

X_model = X.reshape(-1, 1)

model = LinearRegression()
model.fit(X_model, Y)

intercept = model.intercept_
slope = model.coef_[0]

print("\n----- Simple Linear Regression -----")
print("Intercept (b0):", round(intercept, 4))
print("Slope (b1):", round(slope, 4))

print("\nRegression Equation:")
print("Y =", round(intercept, 4), "+", round(slope, 4), "X")

# --------------------------------------------------
# 5. Predictions
# --------------------------------------------------

Y_pred = model.predict(X_model)

print("\n----- Predictions -----")
print("Hours\tActual\tPredicted")

for hours, actual, predicted in zip(X, Y, Y_pred):
    print(f"{hours}\t{actual}\t{predicted:.2f}")

# --------------------------------------------------
# 6. Predict score for a new value
# --------------------------------------------------

new_hours = np.array([[7.5]])
predicted_score = model.predict(new_hours)

print("\n----- New Prediction -----")
print("Hours studied:", new_hours[0][0])
print("Predicted exam score:", round(predicted_score[0], 2))

# --------------------------------------------------
# 7. Model Evaluation
# --------------------------------------------------

r2 = r2_score(Y, Y_pred)
mse = mean_squared_error(Y, Y_pred)
rmse = np.sqrt(mse)

print("\n----- Model Evaluation -----")
print("R-squared:", round(r2, 4))
print("Mean Squared Error:", round(mse, 4))
print("Root Mean Squared Error:", round(rmse, 4))
```
