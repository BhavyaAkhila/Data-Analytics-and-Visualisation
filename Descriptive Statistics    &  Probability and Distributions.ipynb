```python
import numpy as np
import pandas as pd
from scipy.stats import norm


np.random.seed(100)


degradation = np.random.normal(loc=4.5, scale=0.8, size=100)

df = pd.DataFrame({
    "Cell_ID": range(1, 101),
    "Degradation (%)": degradation
})

print("Battery Cell Degradation Data")
print(df.head())


print("\nDescriptive Statistics")
print(df["Degradation (%)"].describe())

mean = np.mean(degradation)
median = np.median(degradation)
std = np.std(degradation)

print("\nMean :", round(mean,2))
print("Median :", round(median,2))
print("Standard Deviation :", round(std,2))



threshold = 6

prob_failure = 1 - norm.cdf(threshold, mean, std)

print("\nProbability of Cell Failure (>6% degradation)")
print(round(prob_failure,4))

expected_failures = prob_failure * len(degradation)

print("Expected Failed Cells :", round(expected_failures))

Z = 1.96
p = 0.5
E = 0.05

sample_size = (Z**2 * p * (1-p)) / (E**2)

print("\nRecommended Sample Size")
print(round(sample_size))

print("\nQuality Assessment Completed Successfully.")
```
